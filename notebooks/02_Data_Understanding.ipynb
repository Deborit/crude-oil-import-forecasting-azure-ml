```python
import pandas as pd
import numpy as np

file_path = r"D:\MBA\Project data\PET_IMPORTS_full.csv"

# ---------------------------------------------------------
# 1. Read only country-level, monthly, all-grades crude oil
# ---------------------------------------------------------

chunks = []

for chunk in pd.read_csv(file_path, chunksize=100000):

    filtered = chunk[
        (chunk["geography2"] == "USA") &
        (chunk["name"].str.contains(
            "all grades of crude oil",
            case=False,
            na=False
        )) &
        (chunk["series_id"].str.startswith(
            "PET_IMPORTS.CTY_",
            na=False
        )) &
        (chunk["series_id"].str.endswith(
            ".M",
            na=False
        ))
    ]

    if not filtered.empty:
        chunks.append(
            filtered[
                [
                    "series_id",
                    "name",
                    "geography",
                    "date",
                    "value"
                ]
            ]
        )

raw_model = pd.concat(chunks, ignore_index=True)

# ---------------------------------------------------------
# 2. Count observations by country
# ---------------------------------------------------------

country_counts = (
    raw_model
    .groupby("geography")["date"]
    .nunique()
    .sort_values(ascending=False)
)

# Keep countries with at least 60 monthly observations
valid_countries = country_counts[
    country_counts >= 60
].index

model_data = raw_model[
    raw_model["geography"].isin(valid_countries)
].copy()

# ---------------------------------------------------------
# 3. Clean date and value
# ---------------------------------------------------------

model_data["date"] = pd.to_datetime(
    model_data["date"].astype(str),
    format="%Y%m"
)

model_data["value"] = pd.to_numeric(
    model_data["value"],
    errors="coerce"
)

model_data = model_data.dropna(
    subset=["date", "value"]
)

model_data = model_data.sort_values(
    ["geography", "date"]
).reset_index(drop=True)

# ---------------------------------------------------------
# 4. Create next-month target
# ---------------------------------------------------------

model_data["target_next_month"] = (
    model_data
    .groupby("geography")["value"]
    .shift(-1)
)

# ---------------------------------------------------------
# 5. Create historical features
# ---------------------------------------------------------

model_data["lag_1"] = (
    model_data
    .groupby("geography")["value"]
    .shift(1)
)

model_data["lag_3"] = (
    model_data
    .groupby("geography")["value"]
    .shift(3)
)

model_data["lag_6"] = (
    model_data
    .groupby("geography")["value"]
    .shift(6)
)

model_data["lag_12"] = (
    model_data
    .groupby("geography")["value"]
    .shift(12)
)

# Rolling averages
model_data["rolling_mean_3"] = (
    model_data
    .groupby("geography")["value"]
    .transform(
        lambda x: x.shift(1).rolling(3).mean()
    )
)

model_data["rolling_mean_6"] = (
    model_data
    .groupby("geography")["value"]
    .transform(
        lambda x: x.shift(1).rolling(6).mean()
    )
)

# Calendar features
model_data["month"] = model_data["date"].dt.month
model_data["quarter"] = model_data["date"].dt.quarter
model_data["year"] = model_data["date"].dt.year

# ---------------------------------------------------------
# 6. Remove rows where lag/target is unavailable
# ---------------------------------------------------------

model_data = model_data.dropna(
    subset=[
        "lag_1",
        "lag_3",
        "lag_6",
        "lag_12",
        "rolling_mean_3",
        "rolling_mean_6",
        "target_next_month"
    ]
).reset_index(drop=True)

# ---------------------------------------------------------
# 7. Save modelling dataset
# ---------------------------------------------------------

output_file = (
    r"D:\MBA\Project data\Petroleum_Import_Forecasting"
    r"\data\crude_import_ml_dataset.csv"
)

model_data.to_csv(
    output_file,
    index=False
)

# ---------------------------------------------------------
# 8. Summary
# ---------------------------------------------------------

print("==============================================")
print("FINAL ML DATASET")
print("==============================================")

print("Countries:", model_data["geography"].nunique())
print("Rows:", len(model_data))
print(
    "Date range:",
    model_data["date"].min(),
    "to",
    model_data["date"].max()
)

print("\nCountries included:")
print(
    model_data["geography"]
    .drop_duplicates()
    .tolist()
)

print("\nSaved to:")
print(output_file)

display(model_data.head(10))
```

```python
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

# ---------------------------------------------------------
# 1. Prepare data
# ---------------------------------------------------------

df = model_data.copy()

# One-hot encode country
df = pd.get_dummies(
    df,
    columns=["geography"],
    dtype=int
)

country_columns = [
    c for c in df.columns
    if c.startswith("geography_")
]

features = [
    "lag_1",
    "lag_3",
    "lag_6",
    "lag_12",
    "rolling_mean_3",
    "rolling_mean_6",
    "month",
    "quarter",
    "year"
] + country_columns

target = "target_next_month"

# ---------------------------------------------------------
# 2. Chronological train/test split
# ---------------------------------------------------------

df = df.sort_values("date").reset_index(drop=True)

cutoff = df["date"].quantile(0.80)

train = df[df["date"] <= cutoff]
test = df[df["date"] > cutoff]

X_train = train[features]
y_train = train[target]

X_test = test[features]
y_test = test[target]

print("==============================================")
print("TRAIN / TEST SPLIT")
print("==============================================")
print("Training rows:", len(train))
print("Testing rows :", len(test))
print("Training end :", train["date"].max())
print("Testing start:", test["date"].min())

# ---------------------------------------------------------
# 3. Train XGBoost
# ---------------------------------------------------------

model = XGBRegressor(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42
)

model.fit(
    X_train,
    y_train
)

# ---------------------------------------------------------
# 4. Predict
# ---------------------------------------------------------

predictions = model.predict(X_test)

# ---------------------------------------------------------
# 5. Evaluate
# ---------------------------------------------------------

mae = mean_absolute_error(
    y_test,
    predictions
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        predictions
    )
)

r2 = r2_score(
    y_test,
    predictions
)

print("\n==============================================")
print("XGBOOST MODEL RESULTS")
print("==============================================")
print("MAE :", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("R²  :", round(r2, 4))
```

```python
# =========================================================
# MODEL COMPARISON
# =========================================================

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor,
    HistGradientBoostingRegressor
)
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np
import pandas as pd
import time

# ---------------------------------------------------------
# Models
# ---------------------------------------------------------

models = {

    "Linear Regression": LinearRegression(),

    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        max_depth=12,
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=200,
        max_depth=5,
        learning_rate=0.05,
        random_state=42
    ),

    "HistGradientBoosting": HistGradientBoostingRegressor(
        max_iter=200,
        learning_rate=0.05,
        max_leaf_nodes=31,
        random_state=42
    ),

    "XGBoost": XGBRegressor(
        n_estimators=300,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        objective="reg:squarederror",
        random_state=42
    )
}

# ---------------------------------------------------------
# Train and evaluate
# ---------------------------------------------------------

results = []

for name, model_test in models.items():

    print(f"Running: {name}")

    start_time = time.time()

    model_test.fit(
        X_train,
        y_train
    )

    pred = model_test.predict(
        X_test
    )

    mae_test = mean_absolute_error(
        y_test,
        pred
    )

    rmse_test = np.sqrt(
        mean_squared_error(
            y_test,
            pred
        )
    )

    r2_test = r2_score(
        y_test,
        pred
    )

    training_time = time.time() - start_time

    results.append({
        "Model": name,
        "MAE": mae_test,
        "RMSE": rmse_test,
        "R2": r2_test,
        "Training_Time_sec": training_time
    })

# ---------------------------------------------------------
# Comparison table
# ---------------------------------------------------------

comparison = pd.DataFrame(results)

comparison = comparison.sort_values(
    "RMSE"
).reset_index(drop=True)

print("\n==============================================")
print("MODEL COMPARISON")
print("==============================================")

display(
    comparison.style.format({
        "MAE": "{:,.2f}",
        "RMSE": "{:,.2f}",
        "R2": "{:.4f}",
        "Training_Time_sec": "{:.2f}"
    })
)
```

```python
# ============================================================
# CREATE CORRECT MASTER PREDICTION DATASET
# ============================================================

# Start from the correctly ordered test dataframe
prediction_results = test[
    [
        "date",
        "value",
        "target_next_month"
    ]
].copy()

# Recover country from the one-hot encoded columns
country_columns = [
    c for c in features
    if c.startswith("geography_")
]

prediction_results["geography"] = (
    test[country_columns]
    .idxmax(axis=1)
    .str.replace("geography_", "", regex=False)
)

# Add predictions from the validated local model
prediction_results["predicted_import"] = y_pred_local

# Rename columns
prediction_results = prediction_results.rename(
    columns={
        "value": "current_import",
        "target_next_month": "actual_next_month"
    }
)

# Calculate prediction error
prediction_results["error"] = (
    prediction_results["actual_next_month"]
    - prediction_results["predicted_import"]
)

prediction_results["absolute_error"] = (
    prediction_results["error"].abs()
)

# Arrange columns
prediction_results = prediction_results[
    [
        "date",
        "geography",
        "current_import",
        "actual_next_month",
        "predicted_import",
        "error",
        "absolute_error"
    ]
]

print("Master prediction dataset created!")
print("Rows:", len(prediction_results))
print("Columns:", len(prediction_results.columns))

print(
    "\nPeriod:",
    prediction_results["date"].min(),
    "to",
    prediction_results["date"].max()
)

display(prediction_results.head(10))
```
