```python
import pandas
import numpy
import sklearn
import statsmodels
import xgboost

print("All ML libraries installed successfully")
import json

file_path = r"D:\MBA\Project data\Unziped_PET_IMPORTS\PET_IMPORTS.txt"

target_id = "PET_IMPORTS.REG_CA-US-ALL.M"

target = None

with open(file_path, "r", encoding="utf-8") as f:
    for line in f:
        if line.strip():
            record = json.loads(line)

            if record.get("series_id") == target_id:
                target = record
                break

if target:
    print("Series found!")
    print("Series ID:", target["series_id"])
    print("Name:", target["name"])
    print("Frequency:", target["f"])
    print("Units:", target["units"])
    print("Start:", target["start"])
    print("End:", target["end"])
    print("Number of observations:", len(target["data"]))
else:
    print("Series not found")

import pandas as pd

df = pd.DataFrame(
    target["data"],
    columns=["date", "import_volume"]
)

# Convert YYYYMM into a proper date
df["date"] = pd.to_datetime(
    df["date"].astype(str),
    format="%Y%m"
)

# Sort chronologically
df = df.sort_values("date").reset_index(drop=True)

# Display basic information
print("Number of rows:", len(df))
print("First date:", df["date"].min())
print("Last date:", df["date"].max())

df.head()

print("----- DATA QUALITY CHECK -----")

print("\n1. Number of rows:", len(df))

print("\n2. Missing values:")
print(df.isnull().sum())

print("\n3. Duplicate dates:", df["date"].duplicated().sum())

print("\n4. Date range:")
print(df["date"].min(), "to", df["date"].max())

print("\n5. Data types:")
print(df.dtypes)

print("\n6. Basic statistics:")
print(df["import_volume"].describe())
```
