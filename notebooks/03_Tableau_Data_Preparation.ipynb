```python
import pandas as pd

file_path = r"D:\MBA\Project data\PET_IMPORTS_full.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Rows:", len(df))
print("Columns:", len(df.columns))

df.head()
```

```python
print("=" * 70)
print("DESTINATION GEOGRAPHIES")
print("=" * 70)

print(df["geography2"].value_counts().to_string())

print("\n" + "=" * 70)
print("SERIES TYPE PATTERNS")
print("=" * 70)

print("\nRefinery-level series:")
print(
    df[df["series_id"].str.contains("-RF_", na=False)]
    [["series_id", "name", "geography", "geography2"]]
    .drop_duplicates()
    .head(20)
    .to_string(index=False)
)

print("\nPoint/terminal-level series:")
print(
    df[df["series_id"].str.contains("-PT_", na=False)]
    [["series_id", "name", "geography", "geography2"]]
    .drop_duplicates()
    .head(20)
    .to_string(index=False)
)

print("\nUSA-level series:")
print(
    df[df["geography2"] == "USA"]
    [["series_id", "name", "geography", "geography2"]]
    .drop_duplicates()
    .head(20)
    .to_string(index=False)
)
```

```python
print("=" * 70)
print("SERIES TYPE COUNTS")
print("=" * 70)

print(
    df["series_id"]
    .str.extract(r"PET_IMPORTS\.([A-Z]+)_")[0]
    .value_counts()
)

print("\n" + "=" * 70)
print("COUNTRY-LEVEL EXAMPLES")
print("=" * 70)

country_examples = (
    df[df["series_id"].str.contains(r"\.CTY_", regex=True, na=False)]
    [["series_id", "name", "geography", "geography2"]]
    .drop_duplicates()
    .head(10)
)

print(country_examples.to_string(index=False))

print("\n" + "=" * 70)
print("REFINERY-LEVEL EXAMPLES")
print("=" * 70)

refinery_examples = (
    df[df["series_id"].str.contains(r"-RF_", regex=True, na=False)]
    [["series_id", "name", "geography", "geography2"]]
    .drop_duplicates()
    .head(10)
)

print(refinery_examples.to_string(index=False))

print("\n" + "=" * 70)
print("POINT / TERMINAL-LEVEL EXAMPLES")
print("=" * 70)

point_examples = (
    df[df["series_id"].str.contains(r"-PT_", regex=True, na=False)]
    [["series_id", "name", "geography", "geography2"]]
    .drop_duplicates()
    .head(10)
)

print(point_examples.to_string(index=False))
```

```python
import re

# Get unique series definitions
series_df = df[
    ["series_id", "name", "geography", "geography2"]
].drop_duplicates()

# Extract the part between the source geography and crude grade
series_df["destination_code"] = (
    series_df["series_id"]
    .str.extract(r"-([A-Z0-9_]+)-ALL\.M$")[0]
)

print("UNIQUE DESTINATION CODE PATTERNS")
print("=" * 70)

print(
    series_df["destination_code"]
    .value_counts()
    .head(50)
    .to_string()
)

print("\n" + "=" * 70)
print("DESTINATION CODE EXAMPLES")
print("=" * 70)

for pattern in ["US", "RF_", "RS_", "PT_"]:
    print(f"\n--- {pattern} ---")

    temp = series_df[
        series_df["destination_code"].str.startswith(
            pattern,
            na=False
        )
    ]

    print(
        temp[
            ["series_id", "name", "geography", "geography2", "destination_code"]
        ]
        .head(10)
        .to_string(index=False)
    )
```

```python
# Create the main Tableau dataset
# Grain: Source Country × Month
# Destination: Total U.S.
# Product: All Grades of crude oil
# Frequency: Monthly
# Source: Country-level

tableau_main = df[
    (df["geography2"] == "USA") &
    (df["series_id"].str.startswith("PET_IMPORTS.CTY_", na=False)) &
    (df["series_id"].str.endswith(".M", na=False)) &
    (df["name"].str.contains(
        "all grades of crude oil",
        case=False,
        na=False
    ))
].copy()

print("Tableau main dataset created.")
print("Rows:", len(tableau_main))
print("Columns:", len(tableau_main.columns))

tableau_main.head()
```

```python
# Create Tableau-friendly columns

tableau_main = tableau_main.copy()

# -----------------------------
# 1. Rename existing columns
# -----------------------------

tableau_main = tableau_main.rename(
    columns={
        "geography": "Source Country Code",
        "geography2": "Destination Country Code",
        "value": "Import Volume (000 Barrels)",
        "units": "Unit"
    }
)

# -----------------------------
# 2. Create readable country fields
# -----------------------------

country_mapping = {
    "AGO": "Angola",
    "ARE": "United Arab Emirates",
    "ARG": "Argentina",
    "AZE": "Azerbaijan",
    "BRA": "Brazil",
    "CMR": "Cameroon",
    "COG": "Republic of Congo",
    "COL": "Colombia",
    "DZA": "Algeria",
    "ECU": "Ecuador",
    "GAB": "Gabon",
    "GHA": "Ghana",
    "GNQ": "Equatorial Guinea",
    "GTM": "Guatemala",
    "GUY": "Guyana",
    "IDN": "Indonesia",
    "IRQ": "Iraq",
    "KAZ": "Kazakhstan",
    "KWT": "Kuwait",
    "LBY": "Libya",
    "MEX": "Mexico",
    "NGA": "Nigeria",
    "NOR": "Norway",
    "PER": "Peru",
    "RUS": "Russia",
    "SAU": "Saudi Arabia",
    "TCD": "Chad",
    "THA": "Thailand",
    "TTO": "Trinidad and Tobago",
    "VEN": "Venezuela",
    "VNM": "Vietnam",
    "CAN": "Canada",
    "GBR": "United Kingdom"
}

tableau_main["Source Country"] = (
    tableau_main["Source Country Code"]
    .map(country_mapping)
    .fillna(tableau_main["Source Country Code"])
)

# -----------------------------
# 3. Destination
# -----------------------------

tableau_main["Destination Country"] = "USA"

tableau_main["Destination Type"] = "Total U.S."

# -----------------------------
# 4. Crude Grade
# -----------------------------

tableau_main["Crude Grade"] = "All Grades"

# -----------------------------
# 5. Convert YYYYMM to Date
# -----------------------------

tableau_main["Date"] = pd.to_datetime(
    tableau_main["date"].astype(str),
    format="%Y%m"
)

# -----------------------------
# 6. Create time dimensions
# -----------------------------

tableau_main["Year"] = tableau_main["Date"].dt.year

tableau_main["Quarter"] = (
    "Q" + tableau_main["Date"].dt.quarter.astype(str)
)

tableau_main["Month Number"] = (
    tableau_main["Date"].dt.month
)

tableau_main["Month Name"] = (
    tableau_main["Date"].dt.month_name()
)

tableau_main["Year-Month"] = (
    tableau_main["Date"].dt.strftime("%Y-%m")
)

# -----------------------------
# 7. Monthly frequency
# -----------------------------

tableau_main["Frequency"] = "Monthly"

print("Tableau-friendly columns created successfully!")

print("\nColumns:")
print(tableau_main.columns.tolist())

print("\nShape:")
print(tableau_main.shape)

tableau_main.head()
```
